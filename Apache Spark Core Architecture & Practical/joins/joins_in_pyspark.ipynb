```python
from pyspark.sql import SparkSession
```

```python
spark=SparkSession.builder.appName('Joins').getOrCreate()
```

```python
emp_data=[
    (1,"Gourav",100),
    (2,"Khushi",101),
    (3,"Divya",102),
    (4,"Rohit",103),
    (5,"Rohini",104)
]
emp_column=["emp_id","emp_name","dept_id"]
```

```python
emp_df=spark.createDataFrame(emp_data,emp_column)
```

```python
dept_data=[
    (100,"HR"),
    (101,"Sales"),
    (102,"Marketing"),
    (103,"Creative"),
    (104,"Devlopment"),
]

dept_column=["dept_id","dept_name"]

dept_df=spark.createDataFrame(dept_data,dept_column)
```

###INNER JOIN

Use the .join() function with  syntax df1.join(df2,condition,how)

```python
# inner_joined = emp_df.join(dept_df,emp_df["dept_id"] == dept_df["dept_id"], "inner")

# ! for remove duplicate value
inner_joined=emp_df.join(dept_df,on="dept_id",how="inner")
inner_joined.show()
```

```python
# left_joined = emp_df.join(dept_df,emp_df["dept_id"] == dept_df["dept_id"], "left")

# ! for remove duplicate value
left_joined=emp_df.join(dept_df,on="dept_id",how="left")
left_joined.show()
```

###FULL OUTER

```python
full_outer=emp_df.join(dept_df,on="dept_id",how="full")
full_outer.show()
```
